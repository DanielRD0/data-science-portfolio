# Componenetes principales de la creacion de un ML(Machine Learnig)
# #1: Los Datos, La materia prima, sin Datos, no hay apredizaje
# #2: Caracteristicas. Atributos y variables que describen cada dato(por separado).
# #3: Modelo: Un algoritmo matematico que aprende patrones
# #4: Entrenamiento: Proceso en el que el modelo ajusta sus parametros
# #5: Evaluacion: Medir que tan bien entrenato y que tan bien comprende la informacion nuestro modelo.

# Libreria: Scikit-Learn
# Scikit-Learn(Sklearn) es una de las bibliotecas mas utilizadas para crear ML en python.

# Tiene un interfaz sencilla y coherente
# Incluye algoritmos listos para usar (Regrecion, Arboles de desicion, Clustering, ETC)
# Permite dividir los datos, entrenar, evaluar y ajustar los modelos en pocas lineas de codigo
# Es compatible con otras librerias como Pandas, Numpy y Matplotlib.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np



# Datos de ejemplo (X: horas de estudio, y: nota esperada)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([50, 60, 65, 70, 80])

# dividir datos
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

#entrenamiento del modelo
modelo = LinearRegression()
modelo.fit(X_train, y_train)

#Evaluar el modelo
Predicciones = modelo.predict(X_test)
error = mean_squared_error(y_test,Predicciones)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Datos de ejemplo (X: horas de estudio, y: nota esperada)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([50, 60, 65, 70, 80])

#entrenamiento del modelo
modelo = LinearRegression()
modelo.fit(X, y)

#Evaluar el modelo
Predicciones = modelo.predict([[6]])
print("La Prediccion de lo generado en el mes 6:",Predicciones)



```
