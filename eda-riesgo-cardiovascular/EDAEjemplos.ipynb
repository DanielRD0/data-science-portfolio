```python
#importar las bibliotecas necesarias
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
#configuracion para visualizaciones
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12
```

```python
#para mostar todas las columnas en DataFrames grandes
pd.set_option('display.max_columns', None)
```

```python
#para reproducibilidad
np.random.seed(42)
```

```python
# Número de registros
n = 1000

# Crear un DataFrame simulado de ventas
data = {
    "fecha": pd.date_range(start="2024-01-01", periods=n),
    "producto_id": np.random.randint(1000, 2000, size=n),
    "categoria": np.random.choice(["Electrónica", "Ropa", "Hogar", "Alimentos", "Deportes"], size=n),
    "precio": np.random.normal(100, 30, size=n).round(2),
    "cantidad": np.random.randint(1, 10, size=n),
    "cliente_id": np.random.randint(5000, 6000, size=n),
    "edad_cliente": np.random.normal(35, 12, size=n).round().astype(int),
    "genero_cliente": np.random.choice(["M", "F", "Otro"], p=[0.48, 0.48, 0.04], size=n),
    "ciudad": np.random.choice(["Madrid", "Barcelona", "Valencia", "Sevilla", "Bilbao"], size=n),
    "metodo_pago": np.random.choice(["Tarjeta", "PayPal", "Transferencia", "Efectivo"], size=n),
    "valoracion": np.random.randint(1, 6, size=n),
    "tiempo_entrega": np.random.normal(5, 2, size=n).round().astype(int)
}
```

```python
# añadir algunos valores nulos para simular datos reales
for clo in ["edad_cliente", "valoracion", "tiempo_entrega"]:
    mask = np.random.rand(n) < 0.05  # 5% de valores nulos
    data[clo] = np.where(mask, np.nan, data[clo])
```

```python
#crear el DataFrame
ventas_df = pd.DataFrame(data)
```

```python
#añadir columnas calculadas
ventas_df["total"] = (ventas_df["precio"] * ventas_df["cantidad"]).round(2)
ventas_df["mes"] = ventas_df["fecha"].dt.month_name()
ventas_df["dia_semana"] = ventas_df["fecha"].dt.day_name()
ventas_df["fin_semana"] = ventas_df["fecha"].dt.dayofweek >= 5  # True si es sábado o domingo
```
