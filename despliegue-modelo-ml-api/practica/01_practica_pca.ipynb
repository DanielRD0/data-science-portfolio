```python
import random
import pandas as pd

positivos = [
"Este producto es excelente y funciona perfectamente",
"La experiencia fue maravillosa y el servicio fue increíble",
"Estoy muy satisfecho con la compra",
"El software funciona rápido y sin errores",
"La comida estaba deliciosa y el servicio fue excelente",
"Me encanta esta aplicación",
"El curso fue muy útil y bien explicado",
"La película fue fantástica",
"La atención al cliente fue excelente",
"El hotel fue muy cómodo y limpio"
]

negativos = [
"El producto llegó dañado",
"La experiencia fue terrible",
"Estoy muy decepcionado con la compra",
"El software tiene muchos errores",
"La comida estaba fría y sin sabor",
"No recomiendo esta aplicación",
"El curso fue confuso",
"La película fue aburrida",
"La atención al cliente fue muy mala",
"El hotel estaba sucio"
]

neutrales = [
"El producto llegó ayer",
"La experiencia fue normal",
"El software fue instalado correctamente",
"La comida fue servida a tiempo",
"El curso tiene varios módulos",
"La película dura dos horas",
"La atención al cliente respondió",
"El hotel está en el centro",
"El pedido llegó en tres días",
"La aplicación tiene varias funciones"
]

data = []

for _ in range(34):
    data.append((random.choice(positivos),"positivo"))

for _ in range(33):
    data.append((random.choice(negativos),"negativo"))

for _ in range(33):
    data.append((random.choice(neutrales),"neutral"))

random.shuffle(data)

df = pd.DataFrame(data, columns=["texto","sentimiento"])

df.head()



```
