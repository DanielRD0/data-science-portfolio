```python
import random
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.decomposition import PCA
```

```python
positivos = [ "Este producto es excelente y funciona perfectamente", "La experiencia fue maravillosa y el servicio fue increíble", "Estoy muy satisfecho con la compra", "El software funciona rápido y sin errores", "La comida estaba deliciosa y el servicio fue excelente", "Me encanta esta aplicación", "El curso fue muy útil y bien explicado", "La película fue fantástica", "La atención al cliente fue excelente", "El hotel fue muy cómodo y limpio" ]
negativos = [ "El producto llegó dañado", "La experiencia fue terrible", "Estoy muy decepcionado con la compra", "El software tiene muchos errores", "La comida estaba fría y sin sabor", "No recomiendo esta aplicación", "El curso fue confuso", "La película fue aburrida", "La atención al cliente fue muy mala", "El hotel estaba sucio" ]
neutrales = [ "El producto llegó ayer", "La experiencia fue normal", "El software fue instalado correctamente", "La comida fue servida a tiempo", "El curso tiene varios módulos", "La película dura dos horas", "La atención al cliente respondió", "El hotel está en el centro", "El pedido llegó en tres días", "La aplicación tiene varias funciones" ]
```

```python
data = []
for _ in range(34):
    data.append((random.choice(positivos),"positivo"))
for _ in range(33):
    data.append((random.choice(negativos),"negativo"))
for _ in range(33):
    data.append((random.choice(neutrales),"neutral"))
random.shuffle(data)
df = pd.DataFrame(data, columns=["texto","sentimiento"])
df.head()
```

```python
#Limpieza de los datos

def limpieza_texto(texto):
 texto = texto.lower()
 texto = re.sub(r'[^\w\s]','',texto)

 return texto

df["texto_arreglado"] = df["texto"].apply(limpieza_texto)
```

```python
#Vectorizacion

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["texto_arreglado"])
Y = df["sentimiento"]
```

```python
#Entrenado del modelo version 1:

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size= 0.3, random_state=42)
```

```python
#Entrenado del modelo version 1:

modelo = LogisticRegression(max_iter=1000)
modelo.fit(X_train, Y_train)

Y_pred = modelo.predict(X_test)
```

```python
#PREGUNTARLE AL PROFE

print(classification_report(Y_test, Y_pred))
```

```python
#Reduccion de dimensionalidad con PCA

pca = PCA(n_components=5)

X_dense = X.toarray()

X_pca = pca.fit_transform(X_dense)
```

```python
#Entrenado del modelo version 2:

X_train_pca, X_test_pca, Y_train, Y_test = train_test_split(X_pca, Y, test_size= 0.3, random_state=42)
```

```python
#Entrenado del modelo version 2:

modelo_pca = LogisticRegression(max_iter=1000)
modelo_pca.fit(X_train_pca, Y_train)

Y_pred_pca = modelo_pca.predict(X_test_pca)
```

```python

print(classification_report(Y_test, Y_pred_pca))
```
