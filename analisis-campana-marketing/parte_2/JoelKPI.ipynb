```

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# ============================
# 1. Cargar los datos
# ============================
df = pd.read_excel("marketing_campaign.xlsx")

# ============================
# 2. Crear columna de gasto total
# ============================
df["GastoTotal"] = (
    df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] +
    df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
)

# ============================
# 3. Scatter plot Ingresos vs Gasto Total con línea de tendencia
# ============================
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="Income", y="GastoTotal", alpha=0.6)
sns.regplot(data=df, x="Income", y="GastoTotal", scatter=False, color="red")
plt.title("Relación entre Ingresos y Gasto Total")
plt.xlabel("Ingresos")
plt.ylabel("Gasto Total")
plt.show()

# ============================
# 4. Histograma de Ingresos
# ============================
plt.figure(figsize=(8,6))
sns.histplot(df["Income"], bins=30, kde=True)
plt.title("Distribución de Ingresos")
plt.xlabel("Ingresos")
plt.ylabel("Frecuencia")
plt.show()

# ============================
# 5. Boxplot de Gasto Total segmentado por quintiles de ingreso
# ============================
df["QuintilIngresos"] = pd.qcut(df["Income"], 5, labels=["Q1","Q2","Q3","Q4","Q5"])

plt.figure(figsize=(8,6))
sns.boxplot(data=df, x="QuintilIngresos", y="GastoTotal")
plt.title("Gasto Total por Quintiles de Ingreso")
plt.xlabel("Quintiles de Ingreso (Q1=Más bajos, Q5=Más altos)")
plt.ylabel("Gasto Total")
plt.show()
```


### Grupo 3 – Relación entre Ingresos y Gasto Total

**Hallazgos:**

1. El *scatter plot* muestra una **tendencia positiva**: a mayores ingresos, mayor gasto total.  
   Sin embargo, no todos los clientes con altos ingresos gastan mucho, lo que indica variabilidad.  

2. El histograma de ingresos revela que la mayoría de los clientes se concentran en ingresos bajos y medios, con menos presencia en los niveles altos.  

3. El boxplot por quintiles de ingreso confirma que el **gasto total aumenta con los ingresos**, pero también se observa más dispersión en los quintiles superiores.  
   Es decir, los clientes con más ingresos no solo gastan más en promedio, sino que también presentan comportamientos de gasto más diversos.  

**Conclusión:**  
Aunque los ingresos son un buen predictor del gasto, no son el único factor.  
Existen segmentos de ingresos bajos que gastan proporcionalmente más, lo que puede representar una oportunidad de segmentación y marketing diferenciado.
