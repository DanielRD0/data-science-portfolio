```python
#Importación de librerias y carga de archivo .CSV
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\jburgos\Downloads\Tarea Data Science\marketing_campaign.csv", sep=";")
pd.set_option('display.max_columns', None)
#Informaciones sobre el DataFrame
df.info()
```

```python
#Creación del columna de gastos totales
df["Gasto_Total"] = df[["MntWines","MntFruits","MntMeatProducts","MntFishProducts","MntSweetProducts","MntGoldProds"]].sum(axis=1)
df["Gasto_Total"]
```

```python
#Generación de diagrama de dispersión.
sns.scatterplot(x=df["Income"], y=df["Gasto_Total"], c="g")
plt.title("Diagrama de Dispersión de Ingresos vs Gastos totales")
plt.ylabel("Gastos Totales")
plt.xlabel("Ingresos")

```

```python
#Identificación de relaciones entre variables.
correlacion = df["Gasto_Total"].corr(df["Income"])

if correlacion == 1:
    print(f"{correlacion} - Correlación positiva perfecta.")

elif  correlacion == -1:
    print(f"{correlacion} - Correlación negativa perfecta.")

elif  correlacion == 0:
    print(f"{correlacion} - Correlación nula.")

elif correlacion >= 0.1 and correlacion <= 0.4:
    print(f"{correlacion} - Correlación debil positiva.")

elif correlacion >= 0.4 and correlacion < 0.7:
    print(f"{correlacion} - Correlación moderada positiva.")

elif correlacion >=0.7 and correlacion < 1:
    print(f"{correlacion} - Correlación alta positiva.")

print("Interpretación: Aunque la tendencia es positiva, no todos los clientes con mayores ingresos gastan proporcionalmente más.")
```
